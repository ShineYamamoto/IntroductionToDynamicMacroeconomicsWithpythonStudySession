## 手順

1. パラメータおよび変数を定義する
$$
\alpha=0.3 , \beta=0.99 , \delta=0.25
$$

2. 以下の方程式を定義する
　\begin{align}
  f(C_{t}, C_{t+1}, K_{t+1})=
   \frac{C_{t+1}}{C_{t}}-\beta(\alpha K_{t+}^{\alpha-1}-\delta+1)=0 \tag{2}\\
  \end{align}
  $$
  g(C_{t}, K_{t}, K_{t+1})=
  K_{t+1}-K_{t}^{\alpha}-(1-\delta)K_{t}+C_{t}=0 \tag{2}\\
  $$
  
  この時、定常状態にあると仮定し$C_{t}=C_{t+1}=C^{*}$としておく（$K$についても同様）

3. (1),(2)式をベクトル化する
  <br>この時のベクトルは、
 $$
        \left[
        \begin{matrix}
            f(C^{*},K^{*})\\
            g(C^{*},K^{*})
        \end{matrix}
        \right]
 $$
  <br>のように記述される。

4. ヤコビ行列Jを作成する
　 <br>ベクトル化された関数に対して、$C^{*},K^{*}$で微分し、
  $$
  \left[\begin{array}{c}
            f_{c}(\bar{C}, \bar{K}) & f_{k}(\bar{C}, \bar{K})\\
            g_{c}(\bar{C}, \bar{K}) & g_{k}(\bar{C}, \bar{K})
        \end{array}\right]
  $$
   <br>のように$2*2$行列にする。

5. ニュートン法を使って定常状態の$C^{*},K^{*}$を近似する
  $$
        \left[\begin{array}{c}
            C^{'}\\
            K^{'}
        \end{array}\right]
        =
        \left[\begin{array}{c}
            \bar{C}\\
            \bar{K}
        \end{array}\right]
        -
        \left[\begin{array}{c}
            f_{c}(\bar{C}, \bar{K}) & f_{k}(\bar{C}, \bar{K})\\
            g_{c}(\bar{C}, \bar{K}) & g_{k}(\bar{C}, \bar{K})
        \end{array}\right]^{-1}
        \left[\begin{array}{c}
            f(\bar{C}, \bar{K})\\
            g(\bar{C}, \bar{K})
        \end{array}\right]
  $$
  収束判定はいろいろと方法はあるが、例えば$C^{'},K^{'}$を(1),(2)式に代入してから合計値を計算し、
  その値が微小な$\epsilon$を下回れば繰り返しを中止する方法がある。
  $\epsilon$の値に関してはpythonの場合計算できる最小値は（たしか）1e-15だった気がするので、まあこれくらいの値でよいかと思われる。

※ 手順1から4まではsympyで実装すると楽にできる。その理由は
　<br>  1. 関数のベクトル化がsympy.Matrix()で簡単にできる
  <br>  2. 関数がfとして定義した場合、f.diff() で微分が簡単にできる
  <br> などがある。
