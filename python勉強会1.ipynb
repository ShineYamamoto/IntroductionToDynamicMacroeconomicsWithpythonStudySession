```python
# パッケージをインストール
import matplotlib.pyplot as plt

# パラメータを定義しておく
α = 0.3
s = 0.2
δ = 0.1
n = -0.01

# 逐次代入をして次期のkを求める
def solow(A, α, s, δ, n, T):   
    # 定常状態のkを求める
    k_star = (s/(n+δ))**(1/(1-α))
    k_init = A * k_star #初期値を設定
    k_list = []
    k_list.append(k_init)

    # 逐次代入して次期のkを求める
    for t in range(T):
        k = ((1-δ)*k_list[t] + s*k_list[t]**α) / (1+n)
        k_list.append(k)
        
    return k_list

T = 100
k_half = solow(0.5, α, s, δ, n, T)
k_double = solow(2, α, s, δ, n, T)

# プロットする
fig = plt.figure()
ax1 = fig.add_subplot(2,2,1)
ax1.plot(k_half)
ax1.set_title(r"$k_{0}=0.5k^{*}$")

ax2 = fig.add_subplot(2,2,2)
ax2.plot(k_double)
ax2.set_title(r"$k_{0}=2k^{*}$")

plt.show()
```
